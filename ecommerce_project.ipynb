```python
import pandas as pd

# רושמים את שם הקובץ האמיתי שלך בתוך המרכאות (כולל הסיומת .csv)
file_name = 'train.csv'

# טעינת קובץ ה-CSV באמצעות המשתנה שהגדרנו למעלה (השם תוקן כאן)
df = pd.read_csv(file_name)

# הצגת 5 השורות הראשונות
display(df.head())
```

```python
# 1. הסרת עמודות המזהים מהטבלה הגולמית שבזיכרון
df_cleaned = df.drop(columns=['Customer_ID', 'Order_ID'])

# 2. דריסה ושמירה מחדש של הקובץ המקורי על הדיסק
# (שים לב: זה ישנה את הקובץ האמיתי, index=False מונע הוספת עמודת מספור שורות מיותרת)
df_cleaned.to_csv('train.csv', index=True)

print("עמודות המזהים הוסרו וקובץ ה-train.csv המקורי עודכן בהצלחה!")
```
